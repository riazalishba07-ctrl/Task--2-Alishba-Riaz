```python
import zipfile
import os

zip_file = "archive (11).zip"

extract_folder = "iris_dataset"

with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall(extract_folder)

print("Dataset Extracted Successfully")
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("iris_dataset/csv_data_iris.csv")

print(data.head())
```

```python
print("========== DATASET INFO ==========")

print(data.info())

print("\n========== DATASET SHAPE ==========")
print(data.shape)

print("\n========== CLASS NAMES ==========")
print(data['class'].unique())
```

```python
X = data.drop(['target', 'class'], axis=1)

y = data['class']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

```python
model = DecisionTreeClassifier()

model.fit(X_train, y_train)

print("Model Trained Successfully")
```

```python
y_pred = model.predict(X_test)

print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy * 100, "%")
```

```python
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
plt.figure(figsize=(6,5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title("Confusion Matrix")

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=data['sepal length (cm)'],
    y=data['petal length (cm)'],
    hue=data['class']
)

plt.title("Iris Flower Classification")

plt.show()
```

```python
sample = [[5.1, 3.5, 1.4, 0.2]]

prediction = model.predict(sample)

print("Predicted Flower:", prediction[0])
```

```python
# Step 16: Display Final Output

print("\n========== FINAL OUTPUT ==========")

# Display Accuracy
print("\nModel Accuracy:")
print(accuracy_score(y_test, y_pred))

# Display Classification Report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Display Confusion Matrix
print("\nConfusion Matrix:")
print(cm)

# Display Final Prediction
print("\nCustom Prediction Result:")
print("Predicted Flower:", prediction[0])

# Show Scatter Plot Again
plt.figure(figsize=(8,6))

sns.scatterplot(
    x=data['sepal length (cm)'],
    y=data['petal length (cm)'],
    hue=data['class']
)

plt.title("Final Iris Flower Classification Visualization")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")

plt.show()

print("\nProject Executed Successfully!")
```
